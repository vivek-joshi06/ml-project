```python
import yfinance as yf
```

```python
import pandas as pd
```

```python
stocks = [
    "RELIANCE.NS",
    "TCS.NS",
    "INFY.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "SBIN.NS",
    "LT.NS",
    "ITC.NS",
    "HINDUNILVR.NS",
    "BAJFINANCE.NS",
    "KOTAKBANK.NS",
    "AXISBANK.NS",
    "BHARTIARTL.NS",
    "MARUTI.NS",
    "SUNPHARMA.NS",
    "TITAN.NS",
    "ULTRACEMCO.NS",
    "ASIANPAINT.NS",
    "WIPRO.NS",
    "ADANIENT.NS"
]

all_data = []

for stock in stocks:
    df = yf.Ticker(stock).history(
        start="2000-01-01",
        end="2025-05-30"
    )

    df.reset_index(inplace=True)
    df["Stock"] = stock

    all_data.append(df)

final_df = pd.concat(all_data, ignore_index=True)

# Put Stock next to Date
final_df = final_df[
    [
        "Date",
        "Stock",
        "Open",
        "High",
        "Low",
        "Close",
        "Volume",
        "Dividends",
        "Stock Splits",
    ]
]

df["Date"] = pd.to_datetime(df["Date"])

print(final_df.head())

final_df.to_csv("stock_price_f.csv", index=False)
```
