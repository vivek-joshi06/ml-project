```python
# 1. Negative price mask
negative_mask = df[['Open', 'High', 'Low', 'Close']].lt(0).any(axis=1)

# 2. OHLC validity mask
invalid_ohlc = (
    (df['High'] < df['Open']) |
    (df['High'] < df['Close']) |
    (df['High'] < df['Low']) |
    (df['Low'] > df['Open']) |
    (df['Low'] > df['Close'])
)

# 3. Invalid OHLC rows that are NOT negative-price rows
invalid_without_negative = df[invalid_ohlc & ~negative_mask]

print("Negative price rows:", negative_mask.sum())
print("Invalid OHLC rows:", invalid_ohlc.sum())
print("Invalid OHLC excluding negative rows:", len(invalid_without_negative))
```
